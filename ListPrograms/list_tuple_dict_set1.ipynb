```python
students = {
    "Rahul": [80, 90, 85],
    "Priya": [95, 92, 98],
    "Amit": [75, 85, 80]
}

subjects=("Math","English","Hindi")

#1 Rahul Math - 80 English - 90  Hindi - 85

for index,name in enumerate(students):
    print(f"{index} {name}",end=" ")

    for sub,stu in zip(subjects,students[name]):
        print(f"{sub}:-{stu}",end=" ")
    print()
    



```

```python
msg = "welcometoitp"

# print(msg.count('e'))

for i in msg:
    print(f"{i}:-{msg.count(i)}")

char_count={}

for char in msg:
    if char in char_count:
        char_count[char]+=1
    else:
        char_count[char]=1
print(char_count)


# character count dictionary

# char_count={'w':1,'e':2,'l':1.......}
```

```python
# list comprehension,dictionary,tuple,set comprehension

# dictionary comprenhension

char_count1 ={char:msg.count(char) for char in set(msg) }

print(char_count1)
```
