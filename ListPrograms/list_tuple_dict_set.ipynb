```python
# enumerate() function
# The enumerate() function gives you both the index (counter) and the value of each element while looping.

for i in range(10,20):
    print(i)

list1=["John","Scott","Martin","Joe"]
for index,value in enumerate(list1,start=1):
    print(f"{index} :- {value}")
```

```python
# zip - combine two iterables 

fruits=["Banana","Kiwi","Mango","Orange"]
qty =[10,5,50,20]

for f,q in zip(fruits,qty):
    print(f"{f}:-{q}")
    
print(type(zip(fruits,qty)))
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union: ", a | b)
print("Intersection: ", a & b)
print("Difference a - b: ", a - b)
print("Symmetric difference:", a ^ b)
print({1, 2}.issubset(a), a.issuperset({1, 2}))
a.intersection_update(b)
print("a after intersection_update:", a)
```
