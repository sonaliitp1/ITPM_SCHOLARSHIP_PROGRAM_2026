```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union: ", a | b)
print("Intersection: ", a & b)
print("Difference a - b: ", a - b)
print("Symmetric difference:", a ^ b)
print({1, 2}.issubset(a), a.issuperset({1, 2}))
a.intersection_update(b)
print("a after intersection_update:", a)
```

```python
products = {
    "P1": ["Laptop", 50000],
    "P2": ["Mouse", 500],
    "P3": ["Keyboard", 1200]
}

cart = [
    ("P1", 1),
    ("P2", 2),
    ("P3", 1)
]

total_bill = 0

for product_id, quantity in cart:
    name, price = products[product_id]
    amount = price * quantity
    total_bill += amount

    print(name, quantity, "x", price, "=", amount)

print("Total Bill:", total_bill)
```

```python
# enumerate() function
# The enumerate() function gives you both the index (counter) and the value of each element while looping.

names = ["Rahul", "Priya", "Amit"]

for i in range(len(names)):
    print(i, names[i])


names = ["Rahul", "Priya", "Amit"]

for index, name in enumerate(names):
    print(index, name)

names = ["Rahul", "Priya", "Amit"]

for roll_no, name in enumerate(names, start=1):
    print(roll_no, name)
```

```python
students = {
    "Rahul": [80, 90, 85],
    "Priya": [95, 92, 98],
    "Amit": [75, 85, 80]
}

subjects = ("Maths", "Science", "English")

for roll_no, (name, marks) in enumerate(students.items(), start=1):
    print("\nRoll No:", roll_no)
    print("Name:", name)

    for subject, mark in zip(subjects, marks):
        print(subject, ":", mark)
```

```python
students = {
    1: ["Rahul", ("IT", "Developer")],
    2: ["Priya", ("HR", "Manager")],
    3: ["Amit", ("IT", "Tester")],
    4: ["Neha", ("HR", "Recruiter")]
}

departments = {}

for student_id, details in students.items():
    name = details[0]
    department = details[1][0]

    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print(departments)
```
